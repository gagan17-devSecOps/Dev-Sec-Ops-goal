```python

cricket = {"Amit", "Rohan", "Sneha", "Karan"}
football = {"Rohan", "Priya", "Karan", "Vikram"}
 
common = cricket & football
if len(common) == 0:
    print("No common players")
else:
    print("Students playing both sports:")
    for student in common:
        print(student)
```

```python
voter_ids = set()
while True:
    vid = input("Enter Voter ID (type 'exit' to stop): ")
    if vid.lower() == "exit":
        break
    voter_ids.add(vid)
 
if len(voter_ids) > 5:
    print("High turnout")
else:
    print("Low turnout")
print("Unique Voter IDs:", voter_ids)
```

```python
required_skills = {"Python", "SQL", "AWS"}
candidate_skills = {"Python", "SQL", "Linux"}
 
if candidate_skills.issuperset(required_skills):
    print("Candidate Hired")
else:
    print("Missing skills:")
    for skill in required_skills:
        if skill not in candidate_skills:
            print(skill)
```

```python
morning = {101, 102, 103, 104}
evening = {103, 104, 105, 106}
 
one_shift = morning ^ evening
for emp_id in one_shift:
    if emp_id in morning:
        print(f"{emp_id} worked morning only")
    else:
        print(f"{emp_id} worked evening only")
```

```python
sentence = input("Enter a long sentence: ")
vowels_found = set()
for ch in sentence.lower():
    if ch in "aeiou":
        vowels_found.add(ch)
print("Unique vowels used:", vowels_found)
```

```python
groceries = {"Rice": 50, "Wheat": 40, "Dal": 120, "Oil": 150, "Sugar": 45}
 
for item, price in groceries.items():
    if price > 100:
        groceries[item] = price - price * 10 / 100  # 10% discount
 
print("Updated prices (₹):", groceries)
```

```python
text = input("Enter a string: ")
freq = {}
for ch in text:
    if ch in freq:
        freq[ch] += 1
    else:
        freq[ch] = 1
print("Character frequency:", freq)
```

```python
rto_codes = {"MH": "Maharashtra", "DL": "Delhi", "KA": "Karnataka"}
 
while True:
    code = input("Enter state code (or 'quit'): ").strip().upper()
    if code == "QUIT":
        break
    if code in rto_codes.keys():
        print("State:", rto_codes[code])
    else:
        print("Code not found")
        choice = input("Do you want to add it? (yes/no): ").strip().lower()
        if choice == "yes":
            state = input("Enter state name: ")
            rto_codes[code] = state
            print(f"{code} added.")
```

```python
students = {
    "Asha": {"Math": 82, "Science": 78},
    "Rahul": {"Math": 70, "Science": 85},
    "Meera": {"Math": 90, "Science": 88},
}
 
print("Students scoring more than 75 in both subjects:")
for name, marks in students.items():
    if marks["Math"] > 75 and marks["Science"] > 75:
        print(name)
 
```

```python
menu = {"Idli": 40, "Dosa": 60, "Vada Pav": 25, "Biryani": 180, "Tea": 15}
total = 0
 
while True:
    option = input("\nChoose: view menu / order item / bill: ").strip().lower()
    if option == "view menu":
        for dish, price in menu.items():
            print(f"{dish}: ₹{price}")
    elif option == "order item" or option == "order":
        dish = input("Enter item name: ").strip().title()
        if dish in menu:
            total += menu[dish]
            print(f"{dish} added. Running total: ₹{total}")
        else:
            print("Item not on the menu")
    elif option == "bill":
        print(f"Final amount to pay: ₹{total}")
        break
    else:
        print("Invalid option")
```
