```python
numbers = [25, 8, 33, 66, 34, 65, 78, 69, 98, 92]
even_numbers = []
odd_numbers = []
for num in numbers:
    if num % 2 == 0:
        even_numbers.append(num)
    else:
        odd_numbers.append(num)
print  ( "Even numbers:", even_numbers )
print  ( "Odd numbers:", odd_numbers )
```

```python
grocery_list = []
num_items = int(input("How many items do you want to add to your grocery list?"))
for i in range(num_items):
    item = input(f"Enter item {i + 1}: " )
    grocery_list.append(item)
print("\nYour final grocery list: ")
print(grocery_list)    
```

```python
vehicles = ["Truck", "car","Bike","Bus","Scooter"]
search_vehicle = input("Enter a vehicle name to search for: ")
found = False
for i in range(len(vehicles)):
    if vehicles[i].lower() == search_vehicle.lower():
        vehicles[i] = "Electric Vehicle"
        found= True
        break 
        
if not found:
       print("Vehicle not found!")
       
print("\nFinal update list:")
print(vehicles)
```
