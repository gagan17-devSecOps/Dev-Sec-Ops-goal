```python
n = int(input("Enter a number: "))
 
print("Using for loop:")
for i in range(1, 11):
    print(n, "x", i, "=", n * i)
 
print("Using while loop:")
i = 1
while i <= 10:
    print(n, "x", i, "=", n * i)
    i += 1
```

```python
num = int(input("Enter an integer to check for prime: "))
 
if num <= 1:
    print(num, "is not a prime number (must be greater than 1).")
else:
    is_prime = True
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        print(num, "is a prime number.")
    else:
        print(num, "is not a prime number.")
```

```python
print("Leap years from 1850 to 2026:")
for year in range(1850, 2027):
    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
        print(year)
```

```python
number = int(input("Enter a multi-digit number: "))
temp = abs(number)
digit_sum = 0
 
while temp > 0:
    digit = temp % 10
    digit_sum += digit
    temp = temp // 10
 
print("Sum of digits:", digit_sum)
```

```python
secret_number = 42
 
while True:
    guess = int(input("Guess the number: "))
    if guess > secret_number:
        print("Too high!")
    elif guess < secret_number:
        print("Too low!")
    else:
        print("Correct! Congratulations, you guessed it!")
        break
        
```
