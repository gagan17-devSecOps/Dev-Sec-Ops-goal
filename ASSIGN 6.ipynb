```python
odd_tuple = (1, 3, 5)
even_tuple = (2, 4, 6)

combined_tuple = odd_tuple + even_tuple
print(f"Original combined tuple: {combined_tuple}")

while True:
    try:
        user_input = input("\nEnter a multiplier to repeat the tuple (or type 'exit' to quit): ")
        
        if user_input.lower().strip() == 'exit':
            print("Exiting the program. Goodbye!")
            break
            
        multiplier = int(user_input)
        
        if multiplier < 0:
            print("Please enter a non-negative integer.")
            continue
            
        repeated_tuple = combined_tuple * multiplier
        print(f"Resulting sequence (repeated {multiplier} times):")
        print(repeated_tuple)
        
    except ValueError:
        print("Invalid input! Please enter a valid integer or 'exit'.")
```

```python
product_list = []

print("Enter details for 3 products:")

while len(product_list) < 3:
    product_name = input(f"\nEnter name for product {len(product_list) + 1}: ").strip()
    
    try:
        price = float(input(f"Enter price for {product_name} (in ₹): "))
        
        if price < 0:
            print("Price cannot be negative. Please try again.")
            continue
            
        product_tuple = (product_name, price)
        product_list.append(product_tuple)
        
    except ValueError:
        print("Invalid price! Please enter a valid number.")

print(f"\nAll collected products: {product_list}")

# 4. Iterate and filter products costing more than ₹500
print("\nProducts costing more than ₹500:")
for product, price in product_list:
    if price > 500:
        print(f"- {product}: ₹{price:.2f}")
```

```python
blood_groups = ('O+', 'A+', 'B+', 'O+', 'AB+', 'O+', 'A-', 'B+', 'O-', 'O+')

search_group = input("Enter a blood group to search (e.g., O+, A+, B+): ").strip()

count_instances = blood_groups.count(search_group)

if count_instances > 0:
    print(f"Blood group '{search_group}' is available and appears {count_instances} times in the tuple.")
else:
    print(f"Blood group '{search_group}' is not available.")
```

```python
employee_records = (
    (101, "Aarav Sharma", "IT", 85000),
    (102, "Diya Patel", "HR", 60000),
    (103, "Vivaan Das", "IT", 72000),
    (104, "Ananya Reddy", "IT", 95000),
    (105, "Kabir Mehta", "Finance", 110000),
    (106, "Isha Verma", "IT", 78000)
)

total_it_high_salary = 0
found_any = False

print("Employees in IT earning more than ₹75,000:")
print("-" * 45)

for record in employee_records:
    emp_id, name, department, salary = record
    
    if department == "IT" and salary > 75000:
        print(f"ID: {emp_id} | Name: {name} | Salary: ₹{salary:,}")
        
        total_it_high_salary += salary
        found_any = True
    else:
        pass

print("-" * 45)
if found_any:
    print(f"Total Expenditure for this group: ₹{total_it_high_salary:,}")
else:
    print("No employees matched the criteria.")
```
