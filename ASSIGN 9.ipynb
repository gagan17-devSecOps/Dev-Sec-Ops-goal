```python
class Employee:
    def __init__(self, emp_id, name, department):
        self.emp_id = emp_id
        self.name = name
        self.department = department
 
 
e1 = Employee(101, "Amit", "IT")
e2 = Employee(102, "Priya", "HR")
print(e1.emp_id, e1.name, e1.department)
print(e2.emp_id, e2.name, e2.department)
```

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width
 
    def calculate_area(self):
        return self.length * self.width
 
 
r = Rectangle(8, 5)
print("Area:", r.calculate_area())
```

```python
class Student:
    def __init__(self, name, total_marks):
        self.name = name
        self.total_marks = total_marks
 
 
s1 = Student("Rahul", 425)
s2 = Student("Sneha", 468)
 
if s1.total_marks > s2.total_marks:
    print(s1.name, "scored higher.")
elif s2.total_marks > s1.total_marks:
    print(s2.name, "scored higher.")
else:
    print("Both students scored the same.")
```

```python
class Wallet:
    def __init__(self):
        self.balance = 0
 
    def add_money(self, amount):
        self.balance += amount
 
    def spend_money(self, amount):
        self.balance -= amount
 
 
w = Wallet()
w.add_money(1000)
w.spend_money(350)
w.add_money(200)
print("Final balance: ₹", w.balance)
```

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price
 
    def apply_discount(self):
        if self.price > 1500:
            self.price = self.price - (self.price * 10 / 100)
            print(f"New price of {self.name}: ₹{self.price}")
        else:
            print("No discount applicable")
 
 
p1 = Product("Headphones", 2000)
p2 = Product("Mouse", 800)
p1.apply_discount()
p2.apply_discount()
```

```python
class ShoppingCart:
    def __init__(self):
        self.items = []
 
    def add_item(self, item_name):
        self.items.append(item_name)
 
    def display_cart(self):
        print("Items in cart:")
        for item in self.items:
            print("-", item)
 
 
cart = ShoppingCart()
cart.add_item("Rice")
cart.add_item("Milk")
cart.add_item("Sugar")
cart.display_cart()
```

```python
class TrainTicket:
    def __init__(self, passenger_name, age):
        self.passenger_name = passenger_name
        self.age = age
 
 
tickets = []
count = 0
while count < 3:
    name = input(f"Enter name of passenger {count + 1}: ")
    age = int(input("Enter age: "))
    tickets.append(TrainTicket(name, age))
    count += 1
 
print("\nPassenger details:")
for t in tickets:
    print("Name:", t.passenger_name, "| Age:", t.age)
```

```python
class SmartPhone:
    def __init__(self):
        self.battery_level = 100
 
    def check_battery(self):
        if self.battery_level < 20:
            print("Warning: Battery low!")
 
    def play_game(self):
        self.battery_level -= 30
        print("Battery level:", self.battery_level)
        self.check_battery()
 
 
phone = SmartPhone()
phone.play_game()
phone.play_game()
phone.play_game()
```

```python
class LibraryBook:
    def __init__(self, title, author, is_issued=False):
        self.title = title
        self.author = author
        self.is_issued = is_issued
 
    def issue_book(self):
        if not self.is_issued:
            self.is_issued = True
            print("Book issued successfully")
        else:
            print("Book is currently unavailable")
 
    def return_book(self):
        self.is_issued = False
        print("Book returned successfully")
 
 
book = LibraryBook("Wings of Fire", "A.P.J. Abdul Kalam")
book.issue_book()
book.issue_book()
book.return_book()
```
