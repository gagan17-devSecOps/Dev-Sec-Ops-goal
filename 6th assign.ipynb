```python
    cities = ("Pune", "Mumbai", "Nagpur", "Nashik")
    city_list = list(cities)
    for i in range(len(city_list)):
        if city_list[i] == "Mumbai":
            city_list[i] = "Thane"
    cities = tuple(city_list)
    print("Updated tuple:", cities)
```

```python
odds = (1, 3, 5)
evens = (2, 4, 6)
combined = odds + evens
print("Combined tuple:", combined)
while True:
    value = input("Enter a multiplier (0 to stop): ")
    if not value.lstrip("-").isdigit():
        print("Please enter a whole number.")
        continue
    multiplier = int(value)
    if multiplier == 0:
        print("Done.")
        break
    print("Result:", combined * multiplier)
```

```python
    products = []
    count = 0
    while count < 3:
        name = input(f"Product {count + 1} name: ")
        price = float(input(f"Product {count + 1} price (₹): "))
        products.append((name, price))
        count += 1
    print("Products costing more than ₹500:")
    for name, price in products:
        if price > 500:
            print(f"{name} - ₹{price}")
        else:
            print("NONE")
```

```python
    records = (
        (1, "Asha", "IT", 90000),
        (2, "Rahul", "HR", 60000),
        (3, "Meera", "IT", 80000),
        (4, "Vikram", "IT", 70000),
        (5, "Sneha", "Finance", 85000),
    )
    total = 0
    print("IT employees earning more than ₹75,000:")
    for emp_id, name, dept, salary in records:
        if dept == "IT" and salary > 75000:
            print(name)
            total += salary
    print(f"Total salary: ₹{total}")
```
