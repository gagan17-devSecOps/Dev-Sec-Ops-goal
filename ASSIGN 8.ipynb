```python
def greet_users():
    while True:
        name = input("Enter your name (or 'quit' to stop): ").strip()
        if name.lower() == "quit":
            print("Goodbye!")
            break
        print(f"Welcome, {name}!")
 
 
greet_users()
```

```python
def calculate_area(length, width):
    area = length * width
    if length == width:
        print(f"Area of Square is {area}")
    else:
        print(f"Area of Rectangle is {area}")
 
 
calculate_area(5, 5)
calculate_area(4, 7)
calculate_area(10, 10)
 
```

```python
def generate_bill(item_name, price, tax_rate=18):
    total = price + (price * tax_rate / 100)
    print(f"{item_name}: Price = ₹{price}, GST = {tax_rate}%, Total = ₹{total:.2f}")
 
 
items = [("Rice", 600, 18), ("Milk", 60, 5), ("Soap", 45, 18)]
for name, price, tax in items:
    if tax == 18:
        generate_bill(name, price)       
    else:
        generate_bill(name, price, tax)   
```

```python
def find_maximum(numbers_list):
    maximum = numbers_list[0]
    for num in numbers_list:
        if num > maximum:
            maximum = num
    return maximum
 
 
numbers = []
for i in range(5):
    numbers.append(int(input(f"Enter integer {i + 1}: ")))
print("Maximum value:", find_maximum(numbers))
```

```python
def evaluate_performance(marks):
    total = 0
    for m in marks:
        total += m
    average = total / len(marks)
    if average >= 40:
        status = "Pass"
    else:
        status = "Fail"
    return total, average, status
 
 
total, average, status = evaluate_performance([65, 48, 72, 55, 80])
print("Total:", total)
print("Average:", average)
print("Status:", status)
```

```python
def calculate_total_expense(*expenses):
    total = 0
    for e in expenses:
        total += e
    print("Total expense: ₹", total)
    if total > 10000:
        print("Warning: Budget exceeded!")
    else:
        print("Within budget.")
 
 
calculate_total_expense(2000, 3500, 1500)
calculate_total_expense(2500, 1800, 3000, 1200, 2200, 1500)
```

```python
def convert_usd_to_inr(amount):
    return amount * 83
 
 
dollars = [10, 50, 100, 5, 20]
rupees = list(map(convert_usd_to_inr, dollars))
print("INR prices (₹):", rupees)
```

```python
def is_senior_citizen(age):
    if age >= 60:
        return True
    else:
        return False
 
 
ages = [12, 65, 34, 60, 78, 45, 59, 82, 23]
seniors = list(filter(is_senior_citizen, ages))
print("Senior citizen ages:", seniors)
```

```python
def is_even(n):
    return n % 2 == 0
 
 
def square(n):
    return n * n
 
 
nums = [3, 8, 12, 5, 20, 7, 14, 1, 6, 17]
squared_evens = list(map(square, filter(is_even, nums)))
print("Squared even numbers:", squared_evens)
```

```python
def check_balance(balance):
    return balance
 
def deposit(balance, amount):
    return balance + amount
 
 
def withdraw(balance, amount):
    if amount > balance:
        return "Error: Insufficient balance."
    else:
        return balance - amount
 
balance = 5000
 
while True:
    print("\n1. Check Balance\n2. Deposit\n3. Withdraw\n4. Exit")
    choice = input("Enter your choice: ")
 
    if choice == "1":
        print("Balance: ₹", check_balance(balance))
    elif choice == "2":
        amount = float(input("Enter amount to deposit: "))
        balance = deposit(balance, amount)
        print("Deposited. New balance: ₹", balance)
    elif choice == "3":
        amount = float(input("Enter amount to withdraw: "))
        result = withdraw(balance, amount)
        if isinstance(result, str):
            print(result)
        else:
            balance = result
            print("Withdrawn. New balance: ₹", balance)
    elif choice == "4":
        print("Thank you!")
        break
    else:
        print("Invalid choice.")
```
